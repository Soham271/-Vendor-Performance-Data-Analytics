```python
import pandas as pd
import sqlite3

conn=sqlite3.connect('inventory.db')
```

```python
tables=conn.execute("SELECT name FROM sqlite_master WHERE type='table';").fetchall()
for table in tables:
    print(table[0])
    
```

```python
for table in tables:
    count_df = pd.read_sql_query(f"SELECT COUNT(*) AS count FROM {table[0]}", conn)
    print(f"Table: {table[0]}, Count of records: {count_df['count'].iloc[0]}")
```
